Single mode: One number occurs most often → that’s the mode.

Bimodal: Two numbers occur with equal highest frequency → both are modes.

Multimodal: More than two numbers tie for highest frequency → all are modes.

No mode: If all values occur equally → no mode exists.

| **Operation on Every Observation** | **Mean**         | **Variance**          | **Standard Deviation**         |
| ---------------------------------- | ---------------- | --------------------- | ------------------------------ |
| Add **k**                          | Old Mean **+ k** | No Change             | No Change                      |
| Subtract **k**                     | Old Mean **− k** | No Change             | No Change                      |
| Multiply by **k**                  | Old Mean **× k** | Old Variance **× k²** | Old Standard Deviation **× k** |
| Divide by **k**                    | Old Mean **÷ k** | Old Variance **÷ k²** | Old Standard Deviation **÷ k** |

```python
#26
customers=np.array([10,15,20,25,30])

frequenc=np.array([3,5,8,6,3])

#find f*X
mean=np.sum(customers* frequenc)/np.sum(frequenc)
print("mean : ",mean)

variance=(np.sum((customers**2)*frequenc)/np.sum(frequenc)-(mean**2))
print("variance : ",variance)

std=np.sqrt(variance)
print("std : ",std)
```

Step 1 → fx = x × f

Step 2 → fx² = x² × f

Step 3 → Σf, Σfx, Σfx²

Step 4 → Mean = Σfx / Σf

Step 5 → Variance = (Σfx² / Σf) − (Mean)²

Step 6 → Standard Deviation = √Variance

```python
#Q96. A manufacturer claims that the average weight of a packet is 500 g. A sample of 64 packets has:
import math
from scipy.stats import norm

mu0 = 500          # claimed mean
n, xbar, sigma = 64, 495, 20
alpha = 0.05

# H0: mu = 500   H1: mu != 500  (two-tailed)
SE = sigma / math.sqrt(n)
Z = (xbar - mu0) / SE
z_crit = norm.ppf(1 - alpha/2)
p_value = 2 * (1 - norm.cdf(abs(Z)))

print("H0: mu = 500 g   |   H1: mu != 500 g")
print("Z statistic:", Z)
print("Critical Z value (two-tailed, alpha=0.05):", z_crit)
print("p-value:", p_value)

if abs(Z) > z_crit:
    print("Decision: Reject H0 (|Z| > z_critical)")
    print("Conclusion: There is sufficient evidence that the average packet weight differs from 500 g.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is not sufficient evidence that the average packet weight differs from 500 g.")
```
